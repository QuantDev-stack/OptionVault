# OptionVault Quick Start Example

This notebook demonstrates how to load, analyze, and query the **OptionVault** datasets using Python, Pandas, and Polars.

```python
import pandas as pd
import polars as pl
import numpy as np
from pathlib import Path

# Set file paths
cash_file = Path("../../samples/cash/RELIANCE.csv")
opt_file = Path("../../samples/options/NIFTY23JUN2214000CE.csv")
greek_file = Path("../../samples/greeks/NIFTY23JUN2214000CE.csv")
```

## 1. Load Cash/Spot Data with Pandas

```python
df_cash = pd.read_csv(cash_file)
df_cash['datetime'] = pd.to_datetime(df_cash['datetime'])
df_cash.head()
```

## 2. Load Options and Calculate Greeks

```python
df_greeks = pd.read_csv(greek_file)
df_greeks['datetime'] = pd.to_datetime(df_greeks['datetime'])
print(f"Average Implied Volatility: {df_greeks['iv'].mean():.2f}%")
print(f"Average Option Delta: {df_greeks['delta'].mean():.4f}")
```
